### https://en.wikipedia.org/wiki/Wien_approximation

```python
import sympy as s
s.init_printing()
```

```python
β, ν, T = s.symbols('beta, nu, T')
ν, λ, h, c, k_B, T = s.symbols('nu, lambda, h, c, k_B, T', real=True, positive=True)
π = s.pi
```

```python
u_wien = 8*π*h*ν**3/c**3 * 1/(s.exp(-h*ν/(k_B*T)) - 1)
u_wien
```

```python
du_wien = s.simplify(s.diff(u_wien, ν))
du_wien
```

```python
# https://physics.nist.gov/cuu/pdf/all_2014.pdf
```

```python
sol_u_wien = s.solve(du_wien, ν)[0]
sol_u_wien
```

```python
sol_b_ν = sol_u_wien / T
sol_b_ν
```

```python
w_wien = (ν**2*u_wien/c).subs({ν:c/λ})
w_wien
```

```python
dw_wien = s.simplify(s.diff(w_wien, λ))
dw_wien
```

```python
sol_w_wien = s.solve(dw_wien, λ)[0]
sol_w_wien
```

```python
sol_b = sol_w_wien * T
sol_b
```
