### Spectral energy distribution per logarithmic interval

Bir spektral dağılım dalga boyuna göre $E_\lambda$, frekansa göre ise $E_\nu$ ile ifade edilebilir. Aynı enerji elemanı için

$$
E_\lambda\,d\lambda = E_\nu\,|d\nu|
$$

olduğundan, $E_\lambda$ ve $E_\nu$ eğrilerinin maksimumları aynı fiziksel noktaya karşılık gelmek zorunda değildir.

Logaritmik spektral aralık kullanılırsa

$$
d\ln\lambda=\frac{d\lambda}{\lambda},
\qquad
d\ln\nu=\frac{d\nu}{\nu},
$$

ve

$$
\lambda E_\lambda = \nu E_\nu
$$

elde edilir. Bu nicelik, **logaritmik spektral aralık başına enerji dağılımını** gösterir. Bu temsilin avantajı, dalga boyu ve frekans gösterimlerinde aynı fiziksel spektral bölgenin tepe noktası olarak ortaya çıkmasıdır.

### Kaynaklar

1. S. W. Stahler and F. Palla, *The Formation of Stars*, Wiley-VCH (2005). Kitapta $\lambda F_\lambda$, açıkça “flux measured per logarithmic wavelength interval” olarak tanımlanmaktadır.

2. A. Kashlinsky, *Cosmic infrared background and early galaxy evolution*, Physics Reports 409 (2005). Makalede logaritmik dalga boyu aralığı başına dağılım
   $$
   F=\lambda I_\lambda=\nu I_\nu
   $$
   biçiminde tanımlanmaktadır.

3. NASA/LAMBDA, *Color Correction of IRAS Flux Densities*. Kaynakta “flux per logarithmic frequency interval” kavramı açıklanmakta ve sabit logaritmik frekans aralığı başına akı için $f_\nu\propto1/\nu$ ilişkisi verilmektedir.

4. Case Western Reserve University, *Astronomical Magnitude Systems*. Ders notlarında
   $$
   \lambda f_\lambda=\nu f_\nu
   $$
   özdeşliği açıkça verilmektedir.

```python
import sympy as s
s.init_printing()
```

```python
β, ν, T = s.symbols('beta, nu, T')
ν, λ, h, c, k_B, T = s.symbols('nu, lambda, h, c, k_B, T', real=True, positive=True)
π = s.pi
```

```python
u_wien = 8*π*h*ν**3/c**3 * s.exp(-h*ν/(k_B*T))
u_wien, ν * u_wien
```

```python
du_wien = s.simplify(s.diff(ν * u_wien, ν))
du_wien
```

```python
sol_u_wien = s.solve(du_wien, ν)[0]
sol_u_wien
```

```python
sol_b_ν = sol_u_wien / T
sol_b_ν
```

```python
w_wien = (ν**2*u_wien/c).subs({ν:c/λ})
w_wien, λ * w_wien
```

```python
dw_wien = s.simplify(s.diff(λ * w_wien, λ))
dw_wien
```

```python
sol_w_wien = s.solve(dw_wien, λ)[0]
sol_w_wien
```

```python
sol_b = sol_w_wien * T
sol_b
```
